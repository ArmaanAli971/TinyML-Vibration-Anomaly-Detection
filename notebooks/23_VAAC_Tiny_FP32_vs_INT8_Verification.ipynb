```python
from pathlib import Path
import tensorflow as tf
import numpy as np

PROJECT_DIR = Path(
    r"C:\Users\armaa\OneDrive\Desktop\TinyML-Vibration-Anomaly-Detection"
)

INT8_MODEL_PATH = (
    PROJECT_DIR
    / "results"
    / "quantized"
    / "vaac_tiny_int8.tflite"
)

print("=" * 60)
print("STEP 228 — INT8 TFLITE MODEL")
print("=" * 60)

print("Model path:")
print(INT8_MODEL_PATH)

print("\nModel exists:", INT8_MODEL_PATH.exists())

if INT8_MODEL_PATH.exists():
    print("Model size:", INT8_MODEL_PATH.stat().st_size, "bytes")
else:
    print("ERROR: INT8 model not found.")
    
```

```python
interpreter = tf.lite.Interpreter(
    model_path=str(INT8_MODEL_PATH)
)

interpreter.allocate_tensors()

print("INT8 TFLite interpreter initialized successfully.")
```

```python
input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

print("=" * 60)
print("INT8 INPUT DETAILS")
print("=" * 60)

for item in input_details:
    print(item)

print("\n" + "=" * 60)
print("INT8 OUTPUT DETAILS")
print("=" * 60)

for item in output_details:
    print(item)
```

```python
input_tensor = input_details[0]
output_tensor = output_details[0]

print("=" * 60)
print("STEP 228 — TFLITE INTERFACE SUMMARY")
print("=" * 60)

print("\nINPUT")
print("Name       :", input_tensor["name"])
print("Shape      :", input_tensor["shape"])
print("Datatype   :", input_tensor["dtype"])
print("Quantization:", input_tensor["quantization"])

print("\nOUTPUT")
print("Name       :", output_tensor["name"])
print("Shape      :", output_tensor["shape"])
print("Datatype   :", output_tensor["dtype"])
print("Quantization:", output_tensor["quantization"])
```

```python
expected_shape = (1, 12000, 1)

actual_shape = tuple(input_tensor["shape"])

print("Expected input shape:", expected_shape)
print("Actual input shape  :", actual_shape)

if actual_shape == expected_shape:
    print("PASS: Input shape is correct.")
else:
    print("WARNING: Input shape differs from expected.")
```

```python
CLASS_NAMES = [
    "Healthy",
    "Ball",
    "Inner Race",
    "Outer Race"
]

print("Expected classes:", CLASS_NAMES)
print("Number of classes:", len(CLASS_NAMES))

print("\nTFLite output shape:", output_tensor["shape"])
```

```python
print("=" * 60)
print("QUANTIZATION PARAMETERS")
print("=" * 60)

print("INPUT")
print("Scale     :", input_tensor["quantization"][0])
print("Zero point:", input_tensor["quantization"][1])

print("\nOUTPUT")
print("Scale     :", output_tensor["quantization"][0])
print("Zero point:", output_tensor["quantization"][1])
```

```python
model_size_bytes = INT8_MODEL_PATH.stat().st_size

model_size_kb = model_size_bytes / 1024
model_size_mb = model_size_bytes / (1024 * 1024)

print("=" * 60)
print("INT8 TFLITE MODEL SIZE")
print("=" * 60)

print(f"Bytes : {model_size_bytes}")
print(f"KB    : {model_size_kb:.2f}")
print(f"MB    : {model_size_mb:.4f}")
```

```python
print("=" * 70)
print("STEP 228 — INT8 TFLITE MODEL VERIFICATION")
print("=" * 70)

print("Model                  : VAAC-Tiny")
print("Model file             :", INT8_MODEL_PATH.name)
print("Model exists           :", INT8_MODEL_PATH.exists())
print("Interpreter initialized:", interpreter is not None)

print("\nInput shape             :", tuple(input_tensor["shape"]))
print("Input datatype         :", input_tensor["dtype"])
print("Output shape            :", tuple(output_tensor["shape"]))
print("Output datatype        :", output_tensor["dtype"])

print("\nNumber of classes       :", len(CLASS_NAMES))
print("Classes                :", CLASS_NAMES)

print("\nModel size              :", f"{model_size_kb:.2f} KB")

print("\nStep 228 completed.")
```
