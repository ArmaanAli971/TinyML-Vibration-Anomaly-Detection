```python
from pathlib import Path
from scipy.io import loadmat

CWRU_PATH = Path("../data/raw/CWRU")

files = sorted(CWRU_PATH.glob("*.mat"))

print("Number of files:", len(files))

for file in files:
    print(file.name)
```

```python
def get_class(filename):

    if filename.startswith("NORMAL"):
        return "Healthy"

    elif filename.startswith("IR007"):
        return "Inner Race"

    elif filename.startswith("B007"):
        return "Ball"

    elif filename.startswith("OR007"):
        return "Outer Race"

    else:
        return "Unknown"


for file in files:
    print(file.name, "->", get_class(file.name))
```

```python
def get_load(filename):

    suffix = filename.split("_")[-1]
    load = int(suffix.replace(".mat", ""))

    return load
```

```python
for file in files:
    print(
        file.name,
        "->",
        get_class(file.name),
        "->",
        get_load(file.name),
        "HP"
    )
```
