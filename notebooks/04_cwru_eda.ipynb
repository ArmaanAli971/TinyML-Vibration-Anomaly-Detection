```python
import matplotlib.pyplot as plt
from scipy.io import loadmat
import numpy as np

file_path = "../data/raw/CWRU/NORMAL_0.mat"

data = loadmat(file_path)

de_key = [
    key for key in data.keys()
    if "DE_time" in key
][0]

vibration = data[de_key].squeeze()
```

```python
FS = 12000
time = np.arange(FS) / FS
```

```python
plt.figure(figsize=(12, 4))

plt.plot(
    time,
    vibration[:FS]
)

plt.xlabel("Time (s)")
plt.ylabel("Acceleration")

plt.title(
    "CWRU NORMAL_0 - Drive-End Vibration"
)

plt.grid()

plt.show()
```

```python
files_to_plot = {
    "Healthy": "../data/raw/CWRU/NORMAL_0.mat",
    "Inner Race": "../data/raw/CWRU/IR007_0.mat",
    "Ball": "../data/raw/CWRU/B007_0.mat",
    "Outer Race": "../data/raw/CWRU/OR007@6_0.mat"
}
```

```python
for label, file_path in files_to_plot.items():

    data = loadmat(file_path)

    de_key = [
        key for key in data.keys()
        if "DE_time" in key
    ][0]

    vibration = data[de_key].squeeze()

    time = np.arange(FS) / FS

    plt.figure(figsize=(12, 4))

    plt.plot(
        time,
        vibration[:FS]
    )

    plt.xlabel("Time (s)")
    plt.ylabel("Acceleration")

    plt.title(
        f"CWRU {label} - Drive-End Vibration"
    )

    plt.grid()
    plt.show()
```
